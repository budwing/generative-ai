## 9. 大模型评估与测试
本章主要介绍了机器学习评价指标、大模型基准测试及测试框架等内容。

### 9.1  机器学习评价指标
机器学习评价指标是评估模型性能的常用工具，大语言模型的评价与测试也是在这些指标基础之上建立起来的。

#### 9.1.1  Precision和Recall
Precision和Recall最早应用于信息检索领域，后来随着机器学习技术的发展才用于评价模型性能。Precision一般翻译为查准率或精确率，在机器学习领域则更多使用精确率这一称呼。Recall也有两种中文翻译查全率或召回率，在机器学习中更多使用召回率这个名称。二者的计算公式和具体含义请参考书中介绍。

#### 9.1.2  F1和MCC
F1值（F1 Score）是Precision和Recall的调和平均值，在机器学习中更常用。F1值的计算公式和具体含义请参考书中介绍。

#### 9.1.3  MSE和MAE
MSE（Mean Squared Error，均方误差）和MAE（Mean Absolute Error，平均绝对误差）是回归问题的常用评价指标。MSE更关注预测值与真实值的误差平方，而MAE更关注预测值与真实值的误差绝对值，具体计算公式请参考书中介绍。

#### 9.1.4  Pearson和Spearman*
皮尔逊相关系数（Pearson Correlation Coefficient）和斯皮尔曼相关系数（Spearman Rank Correlation Coefficient）衡量模型预测值之间的相互关系或趋势，只要关系或趋势正确就会有较高的评分。它们的计算公式和具体含义请参考书中介绍。

#### 9.1.5  BLEU和ROUGE
BLEU和ROUGE是文本生成任务中最为常用的两种评价指标，BLEU常用于评价机器翻译的质量，而ROUGE则主要用于评价文本摘要和文本生成的质量。具体含义与计算公式请参考书中介绍。

### 9.2  大模型基准测试
本节在介绍大模型基准测试基本概念的基础之上，介绍了几种常见的基准测试。模型评价指标的计算受到许多因素的影响，为了让模型评价指标的比较更加公平合理，需要引入统一的基准测试（Benchmark）。所谓基准就是基于相同的标准，它让不同事物之间的比较具备了基础。基准测试则是建立在统一标准之上的测试，测试结果可以更加公平地比较被测试对象之间的差异。一般来说，基准测试可由测试任务、数据集、评价指标和测试平台等四部分组成。

#### 9.2.1  测试任务与数据集
大模型的基准测试是由一组测试任务组成的，每个任务都是由标准的数据集和评价指标组成的完整测试。基准测试的数据集一般分为训练集、验证集和测试集三个子集，传统基准测试的训练集要远大于验证集和测试集，但大模型基准测试的训练集一般会与验证集和测试集大小相当，有些甚至会小于测试集。这是因为传统基准测试针对的是空白的小模型，而大模型基准测试针对的是经过预训练的大模型，所以大模型的训练集仅用于模型微调。

#### 9.2.2  GLUE概览
GLUE（General Language Understanding Evaluation）由纽约大学、华盛顿大学和DeepMind等机构的研究人员共同开发，是转换器模型诞生以来较早用于大模型的基准测试。GLUE可用于评估大语言模型在自然语言理解（即NLU）任务上的性能，由9种任务组成。这些任务涵盖了文本分类、语义相似度判断、问答等多个方面，具体请参考书中介绍。

#### 9.2.3  GLUE测试任务
GLUE测试任务可分为单语句任务、相似性任务和推理任务三个大类，具体请参考书中介绍。

#### 9.2.4  SuperGLUE概览
GLUE虽然集齐了基准测试的四个要素，是一个非常不错的针对大语言模型的基准测试。但随着大模型技术的进一步发展，越来越多的模型在GLUE基准上表现优秀，GLUE测试也就变得越来越不具备区分度了。在这种背景下，人们又推出了GLUE的继任者SuperGLUE，这是一个包含了更具挑战性任务的基准测试。SuperGLUE包含有4大类的测试任务和8个数据集，具体请参考书中介绍。

#### 9.2.5  MMLU与GAIA
MMLU（Massive Multitask Language Understanding）包含有57个任务，涵盖了人文、社会、数学、科学、工程、医学、法律等诸多领域的知识。MMLU中许多任务的难度不亚于高考，有些甚至需要专家级别的水平才能解答。MMLU是当前衡量大语言模型性能水平的黄金标准之一，顶级大模型的测试结果达到甚至超过了人类专家的水平。除了MMLU以外，GAIA（General AI Assistants）是另一个值得一提的基准测试。如果说MMLU是针对大模型自身的测试，GAIA则可以认为是对第7章介绍的ReAct和Agent等技术的综合测试，具体请参考书中介绍。

### 9.4  本章小结
本章主要集中在大模型性能的评价指标及测试体系上，它们在大模型训练成本极高的背景下变得越来越重要。可用于大模型性能评估的评价指标包括：
- Accuracy/Precision/Recall：适用于分类任务
- F1/MCC：适用于分类任务
- MSE/MAE：适用于回归任务
- Pearson/Spearman：适用于回归任务，关注预测结果的相对误差
- BLEU/ROUGE：适用于生成式任务

为了提供公平公正的评价体系，人们还设计了不少专门用于大模型评估的基准测试。基准测试一般包括测试任务、数据集、评价指标和测试平台四个部分，测试任务及其相关数据集是基准测试中最核心的内容。大模型基准测试相对于普通基准测试具有两个明显的特征，一是大模型基准测试涵盖的范围更广，二是训练集中的数据量极少。所以大模型基准测试更多的是评价模型执行多种任务的能力，以及模型在少量微调、少量提示甚至零提示下的泛化能力。以下是一些常见的基准测试：
- GLUE/SuperGLUE：考查模型对自然语言的理解能力
- MMLU：考查模型的常识知识
- GAIA：处理真实任务的能力和稳健性
- GSM8K：考查模型数据推理能力
- HumanEval：考查模型代码生成能力

本章还介绍了lm-evaluation-harness和HELM两种常用的基准测试框架，通过它们可以使用简单配置就执行各种基准测试。
