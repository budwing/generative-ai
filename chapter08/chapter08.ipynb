## 8. 大模型部署与优化
本章介绍了大模型在部署时一些挑战，包括大模型对存储、算力等基础设施的需求及相关估算方法等等。同时为了提升大模型在部署时的性能，也介绍了一些优化方法，如模型量化、模型剪枝、模型量化、模型量化等等。

### 8.1  重新认识GPU
这一节主要解释了原本用于图像处理的GPU为什么会与人工智能相关，同时还介绍了GPU相关的一些性能指标。

#### 8.1.1  从GPU到GPGPU
GPGPU是通用图形处理器（General-Purpose computing on Graphics Processing Units）的概念，它的目的是将图像并行处理的能力拓展到其它应用领域。NVIDIA在2007年发布了用于通用计算的CUDA（Compute Unified Device Architecture）框架，并将GPU核心改造为支持通用计算的CUDA核心，这让GPGPU一下子就从概念走进了现实。CUDA框架是由C/C++语言编写的并行计算平台和编程模型，它从软件层面支持开发人员将程序运行在GPU的众多核心上；而CUDA核心则是从硬件层面支持CUDA框架的最小计算单元，将GPU核心的功能从传统的图形处理扩展到了通用计算。

#### 8.1.2  算力指标FLOPS
GPU算力水平的主要指标是每秒浮点运算次数（Floating Point Operations Per Second，以下简称FLOPS），它体现的不是GPU单颗核心的算力水平而是整体算力水平。GPU单核核心性能越高、核心数量越多，GPU的FLOPS也就越高。类似于其它计算机度量单位，FLOPS也有K、M、G等不同量级的单位。

#### 8.1.3  显卡指标
GPU不是显卡，只是显卡上的处理器。除了GPU的算力水平，还有显存大小、显存带宽、显存使用率、显存使用效率等等指标。本小节以NVIDIA的各系列显卡为例，介绍其性能指标及其背后的逻辑。

### 8.3  大模型优化技术**
本节介绍了几种大模型优化技术，它们可以在保持模型性能大致不变的前提下有效地缩减大模型的参数规模，从而以节省大量的成本并拓展大模型的适用范围。

#### 8.3.1  知识蒸馏
知识蒸馏也称模型蒸馏（Knowledge Distillation/Model Distillation）[ Hinton et al., 2015, Distilling the Knowledge in a Neural Network]，是一种由大型模型向小型模型传递知识的模型压缩方法。由于该方法就像是大模型在教小模型学习知识一样，所以人们形像地将这两个模型称为教师模型和学生模型。知识蒸馏的具体原理和实现算法请参考书中介绍。

#### 8.3.2  参数剪枝
参数剪枝（Pruning）通常简称为剪枝，它通过删除神经网络中的冗余参数，或是减少不重要的结构来压缩模型。核心思想就是只保留重要的参数或结构，以减少模型在计算和存储上的成本，进而提升模型推理的性能。具体请参考书中介绍。

#### 8.3.3  量化
量化（Quantization）是将模型参数由高精度浮点数转换为低精度数据类型的过程，比如从32位浮点数（FP32）转换为16位浮点数（FP16/BF16）或者8位整数（INT8）。由于低精度数据类型占用更小的存储空间，所以这种转换可以显著减少模型的显存需求并加速推理过程。具体请参考书中介绍。

### 8.4  本章小结
本章从GPU及其相关算力集群入手，介绍了支持大模型训练与推理的算力基础设施。GPU之所以被广泛应用于人工智能领域，是因为它们具有强大的并行运算能力，这刚好与机器学习中大量张量运算相契合。GPU的算力指标为每秒浮点运算次数FLOPS，目前大多数GPU算力水平处于TFLOPS级别；而由GPU组建的算力集群则处于EFLOPS级别，显卡总量可达数千甚至上万块。除了算力水平以外，显存和卡间通信速率对于大模型运算也很重要，它们在很大程度上影响模型的分布式部署。
大模型算力需求估算是本章的重点内容。它基于一个基本的理论假设，即正向传播每个词元在每个参数上的算力需求为2个FLOPS，而反向传播则需要4个FLOPS。由于推理阶段只有正向传播，而训练阶段则同时需要正反两个过程，所以模型推理和训练的算力估算方法并不完全相同。推理阶段的算力需求除了与参数规模正相关以外，与词元生成速率、并发请求数量也是正相关的。在使用KV缓存的情况下虽可显著降低推理算力需求，但会增加推理过程中的显存需求。训练阶段的算力需求则只与参数规模及训练数据量正相关，训练数据量越大所需算力总量也就越大。大模型预训练阶段的算力需求通常采用PFLOP/s-day表示，即1PFLOPS算力集群运算一天所产生的算力。但也有采用卡时作为单位的，比如DeepSeek就以H800的卡时标识预训练算力总量。DeepSeek的出现也打破了前述理论假设，它将模型训练每词元在每个参数的算力需求降至4FLOPS。

大模型显存需求估算也是本章重点，但相对来说比较简单，一般只与参数模型相关。比如加载大模型的最低显存需求，就是单个参数的大小与参数总量的乘积。但由于模型推理与训练过程中都需要额外空间保存临时变量，所以实际显存的需求可能是模型参数数量的几倍。推理阶段的显存需求较小，一般可预留2~3倍的显存空间；训练阶段的显存需求较大，至少要保证有6倍的显存空间才能开始训练。

在介绍算力和显存需求估算方法的同时，本章还介绍了大模型的扩展定律和Chinchilla法则。扩展定律明确地指出，训练大模型的计算量、数据集和参数规模大小，与大模型性能呈幂律关系。这说明在模型性能相对较高情况下，如果想要继续提升模型性能，就需要在模型规模上加大投入力度。这也是近些年来人们在模型规模上持续投入的理论依据。扩展定律还指出参数规模与训练数据量存在依赖关系，当参数总量增加8倍时，训练数据集需要相应增加5倍。但Chinchilla法则在复现该估算时发现，当训练数据量达到参数规模的20倍时，模型性能才能达到最优。这些经验数据为人们估算模型参数、训练数据及训练成本，提供了非常重要的参考。

本章最后还介绍了三种大模型优化方法，它们是知识蒸馏、参数剪枝和参数量化。知识蒸馏采用大模型训练小模型的方法，将知识从大模型向小模型迁移。本质上来说，知识蒸馏属于训练模型的一种方法，目前已经被广泛应用于大模型的训练与微调中。参数剪枝则是去除模型中影响较小的参数，可以去除单个参数也可以是去除一个单元。参数剪枝会或多或少影响模型性能，有可能需要在剪枝后对模型进行二次微调。量化也是目前应用比较广泛的技术，它的基本思想就是用低精度取代高精度，从而降低模型对算力和存储的需求。但量化并非简单的强制类型转换，而将高精度数据等比例地压缩到低精度空间中，并且在运算过程中还要还原为原始精度。在实践中，这些模型优化方法常常结合在一起使用，实现模型对算力及存储资源的更高效利用。
